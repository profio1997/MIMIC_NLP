```python
PATH_D_ICD = 'DIAGNOSES_ICD.csv.gz'

df_diagnoses_icd = pd.read_csv(PATH_D_ICD, compression='gzip',
                               dtype={'SEQ_NUM': 'Int64'})
df_diagnoses_icd.info()
df_diagnoses_icd.iloc[0]
```

```python
ICD_FILTER = [
    '4019'
]
df_filtered = df_diagnoses_icd[df_diagnoses_icd['ICD9_CODE'].isin(ICD_FILTER)]
subj_filtered = df_filtered['ROW_ID'].unique()
print(f"Number of ICD-9 {ICD_FILTER} patients: {subj_filtered.size}")
```

```python
PATH_TO_NOTEEVENTS = 'NOTEEVENTS.csv.gz'

df_noteevents = pd.read_csv(PATH_TO_NOTEEVENTS, compression='gzip')
df_noteevents.info()
df_noteevents.iloc[0]
```

```python
print(df_noteevents.iloc[0]['TEXT'])
```

```python
df_noteevents["CHARTTIME"] = df_noteevents["CHARTTIME"].fillna(df_noteevents["CHARTDATE"] + " 00:00:00")

df_noteevents["STORETIME"] = pd.to_datetime(df_noteevents["STORETIME"], errors="coerce")
df_noteevents['CHARTTIME']
```

```python
df_noteevents_filtered = df_noteevents[df_noteevents['SUBJECT_ID'].isin(subj_filtered)]
df_noteevents_filtered.info()
```

```python
df_noteevents_filtered['CATEGORY'].unique()
```

```python
df_4019_nutrition_notes = df_noteevents_filtered[df_noteevents_filtered['CATEGORY'] == 'Nutrition']
df_4019_nutrition_notes.info()
```

```python
print(df_4019_nutrition_notes.iloc[0]['TEXT'])
```

```python
df_4019_nutrition_notes.to_csv('df_4019_nutrition_notes.csv', index=False)
```

```python
output_path = "df_4019_nutrition_notes.txt"

with open(output_path, "w", encoding="utf-8") as fout:
    for note in df_4019_nutrition_notes["TEXT"]:
        # write the note
        fout.write(note.strip())
        # add two newlines as a separator
        fout.write("\n\n")

print(f"Wrote {len(df_4019_nutrition_notes)} notes to {output_path}")
```

```python
output_path = "df_4019_nutrition_notes_limited.txt"
max_notes   = 25

with open(output_path, "w", encoding="utf-8") as fout:
    # iterate over at most max_notes rows
    for note in df_4019_nutrition_notes["TEXT"].iloc[:max_notes]:
        fout.write(note.strip())
        fout.write("\n\n")

print(f"Wrote {min(len(df_4019_nutrition_notes), max_notes)} notes to {output_path}")
```
